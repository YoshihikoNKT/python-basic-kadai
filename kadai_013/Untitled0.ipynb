```python
def calculate_price_with_tax(price, tax_rate):
    """
    商品の金額に消費税を加えた合計金額を計算する関数

    :param price: 商品の金額
    :param tax_rate: 消費税率（%）
    :return: 消費税を加えた合計金額
    """
    tax = price * (tax_rate / 100)
    total = price + tax
    return total

# 第1引数に商品の金額「110」、第2引数に消費税率「10」を設定して呼び出す
result = calculate_price_with_tax(110, 10)
print(result)
```
